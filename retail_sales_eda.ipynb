# Exploratory Data Analysis — Retail Sales
This notebook performs data inspection, cleaning, descriptive statistics, grouping, correlation analysis, and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv('../data/raw_retail_sales.csv')
df.head()
```

```python
print(df.shape)
print(df.info())
print('\nMissing values:\n', df.isna().sum())
print('\nDuplicates:', df.duplicated().sum())
```

```python
df = df.drop_duplicates().copy()
for col in ['Marketing_Spend','Avg_Product_Price']:
    df[col] = df[col].fillna(df[col].median())
df['Date'] = pd.to_datetime(df['Date'])
df.describe()
```

```python
print(df.groupby('Product')['Sales'].sum().sort_values(ascending=False))
print(df.groupby('Region')['Sales'].sum().sort_values(ascending=False))
```

```python
monthly = df.groupby('Date', as_index=False)['Sales'].sum()
plt.figure(figsize=(9,5)); plt.plot(monthly['Date'], monthly['Sales'], marker='o'); plt.title('Monthly Sales Trend'); plt.xlabel('Date'); plt.ylabel('Sales'); plt.tight_layout(); plt.show()
```
